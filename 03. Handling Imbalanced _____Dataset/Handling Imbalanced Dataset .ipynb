## 1.import necessary libraries

```python
from sklearn import datasets
import pandas as pd
```

## 2.import dataset

```python
braest_cancer_data = datasets.load_breast_cancer()
```

```python
braest_cancer_data
```

```python
pd.DataFrame(data = braest_cancer_data.data,columns = braest_cancer_data.feature_names)
```

```python
print(braest_cancer_data.DESCR)
```

```python
braest_cancer_data_df = pd.DataFrame(data = braest_cancer_data.data,columns = braest_cancer_data.feature_names)
braest_cancer_data_df["Target"] = braest_cancer_data.target
braest_cancer_data_df
```

## 3.Data understanding

## 4.model building

```python
x = braest_cancer_data_df.drop(labels = ["Target"],axis = 1)
y = braest_cancer_data_df["Target"]                            
```

```python
x.shape,y.shape
```

```python
y.value_counts()
```

```python
from matplotlib import pyplot as plt
```

```python
plt.pie(x = y.value_counts(),labels = ["1 -  benign ","0 -  Malignant"],explode = [0.01,0.01],autopct = ("%1.0f%%"))
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x,y, test_size = 0.20, random_state = 12, shuffle=True, stratify=y)
```

```python
print("tainning samples",X_train.shape,y_train.shape)
print("testing samples",X_test.shape,y_test.shape)
```

```python
plt.pie(x = y_train.value_counts(),labels = ["1 -  benign ","0 -  Malignant"],explode = [0.01,0.01],autopct = ("%1.0f%%"))
plt.show()
```

```python
plt.pie(x = y_test .value_counts(),labels = ["1 -  benign ","0 -  Malignant"],explode = [0.01,0.01],autopct = ("%1.0f%%"))
plt.show()
```

## 5.model Trainig

```python
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(max_depth = 3,criterion = 'gini',class_weight = {0:2, 1:1} )
dt_model.fit(X_train,y_train)
```

## 6.model Testing || 7. model Evaluation

### Training Data 

```python
y_pred_train = dt_model.predict(X_train)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_train,y_pred_train))
```

### test data 

```python
y_pred_test = dt_model.predict(X_test)
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```
